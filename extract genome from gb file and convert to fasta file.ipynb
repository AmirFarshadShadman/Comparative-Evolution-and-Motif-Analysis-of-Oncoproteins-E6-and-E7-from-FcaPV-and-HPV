```python
from pathlib import Path
from Bio import SeqIO
from Bio.SeqRecord import SeqRecord

# =====================================================
# 
# =====================================================

INPUT_FOLDER = Path(r"E:")          
OUTPUT_FASTA = Path(r"E:...\all_genomes.fasta")  

# =====================================================
# 
# =====================================================

records = []
processed = 0
failed = 0


gb_files = sorted(list(INPUT_FOLDER.glob("*.gb")) +
                  list(INPUT_FOLDER.glob("*.gbk")))

if not gb_files:
    print("not found.")

for gb_file in gb_files:
    try:
        
        record = SeqIO.read(gb_file, "genbank")

        # 
        header = gb_file.stem

        # 
        fasta_record = SeqRecord(
            seq=record.seq,
            id=header,
            name=header,
            description=""
        )

        records.append(fasta_record)
        processed += 1

        print(f"✓ {gb_file.name}  -->  {header}   ({len(record.seq):,} bp)")

    except Exception as e:
        failed += 1
        print(f"✗ error {gb_file.name}")
        print(e)

# 
SeqIO.write(records, OUTPUT_FASTA, "fasta")

print("\n" + "=" * 60)
print(f"input number : {len(gb_files)}")
print(f"success                : {processed}")
print(f"fail             : {failed}")
print(f"saved count: {len(records)}")
print(f"output:")
print(OUTPUT_FASTA)
print("=" * 60)
```
